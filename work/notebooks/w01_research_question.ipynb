## 2. The question: decision, action, cost of a wrong call

* **Decision Improved:** Identifying which existing blog posts or articles should be flagged for immediate editorial refresh versus left untouched.
* **Who Acts:** Content Strategists, SEO Managers, and Editorial Writers.
* **Cost of a Wrong Call:**
  * **False Positive (Flagging a healthy page):** Wasted writer hours and editorial budget rewriting content that was already performing well, risking a loss in existing search rankings.

## 3. Quick look at the data (2-3 real numbers)

### Key Empirical Findings from the Starter Data:

1. **0.001 Correlation (Search Volume vs. Actual Impressions):** Keyword search volume has virtually zero linear correlation with actual page impressions, proving that target search volume is a misleading proxy for prioritization.
2. **57.2% Declining Rate for Stale Pages:** Pages that have not been updated in over 180 days show a 57.2% probability of declining traffic, confirming staleness as a viable candidate signal.
3. **2,909 vs 2,912 Median Word Count:** Declining pages have nearly the exact same median word count as stable pages, proving that content length alone is not the lever for performance.

```python
# 1. Search Volume vs Impressions Correlation
corr_sv_imp = df["search_volume"].corr(df["impressions_90d"])

# 2. Proportion of stale pages (>= 180 days) that are declining
stale_pages = df[df["days_since_last_update"] >= 180]
stale_declining_rate = (stale_pages["trend_direction"] == "down").mean()

# 3. Median word count comparison
med_wc_down = df[df["trend_direction"] == "down"]["word_count"].median()
med_wc_stable = df[df["trend_direction"] == "stable"]["word_count"].median()

print(f"1. Correlation (Search Volume vs. Impressions): {corr_sv_imp:.3f}")
print(f"2. Declining rate among stale content (>= 180 days): {stale_declining_rate:.1%}")
print(f"3. Median Word Count — Declining: {med_wc_down:.0f} vs. Stable: {med_wc_stable:.0f}")
```
